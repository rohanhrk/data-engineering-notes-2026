```python
from pydantic import BaseModel
```

```python
class employee_schema(BaseModel):
    name: str
    id: int
    department: str
```

```python
obj = employee_schema(name="Rohan", id=1, department="Data Engineering")
obj
```

```python
obj2 = employee_schema(**{"name": "Rohan", "id":1, "department":"Data Engineering"})    
obj2
```

# **Coercion VS Strict Types**



## **COERCION**


```python
obj3 = employee_schema(name="Rohan", id="101", department="Data Engineering") # "101" can be converted to int automatically by Pydantic
obj3
```

## **StrictTypes**

```python
from pydantic import StrictInt
```

```python
class employee_schema(BaseModel):
    name: str
    id: StrictInt
    department: str 
```

```python
obj4 = employee_schema(name="Rohan", id="101", department="Data Engineering") # "101" cannot be converted to int automatically by Pydantic because we are using StrictInt
obj4
```

```python
obj5 = employee_schema(name="Rohan", id=101, department="Data Engineering") # "101" is an int, so it will work with StrictInt
obj5
```
